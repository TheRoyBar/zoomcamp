```python
import yfinance as yf
import pandas as pd
import numpy as np
```

```python
# Downloading dataset from yfinance
tickers = {
    'United States': '^GSPC',
    'China': '000001.SS',
    'Hong Kong': '^HSI',
    'Australia': '^AXJO',
    'India': '^NSEI',
    'Canada': '^GSPTSE',
    'Germany': '^GDAXI',
    'United Kingdom': '^FTSE',
    'Japan': '^N225',
    'Mexico': '^MXX',
    'Brazil': '^BVSP',
}

start_date = '2026-01-01'
end_date = '2026-08-21'

data = yf.download(list(tickers.values()), start=start_date, end=end_date)['Close']
```

```python
data
```

YTD Formula = ((Current Value - Beginning Value)/ Beginning Value) *100

```python
ytd_returns = {}
```

```python
for country, ticker in tickers.items(): # Goes through each country and ticker
    series = data[ticker].dropna() # Selects the particular ticker and saves it in series

    # Using iloc to get first and last location in the series
    first_close = series.iloc[0]
    last_close = series.iloc[-1]

    # Applying the ytd formula
    ytd_return = (last_close - first_close) / first_close * 100
    ytd_returns[country] = ytd_return # Saving each in a list
```

```python
ytd_df = pd.Series(ytd_returns).sort_values(ascending=False)
```

```python
ytd_df
```

```python
#For number of countries better than the United States:

us = ytd_returns['United States']
better_than_us = ytd_df[ytd_df > us]

print(f"\nIndexes with better YTD returns than the US ({len(better_than_us)} out of 10):")
print(better_than_us)
```

```python
import pandas as pd
import numpy as np
import requests
from io import StringIO
```

```python
url = 'https://en.wikipedia.org/wiki/List_of_S%26P_500_companies'
headers = {'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome'}

response = requests.get(url, headers=headers)
tables = pd.read_html(StringIO(response.text))

sp500 = tables[0]
```

```python
sp500
```

```python
#Creating the required Dataframe.

df1 = sp500[['Symbol', 'Security', 'Date added']]
df1
```

```python
df1['Date added'].dtype
```

```python
df1['Date'] = pd.to_datetime(df1['Date added'])
```

```python
df1 =df1.drop(columns='Date added')
```

```python
df1['Year'] = df1['Date'].dt.year
```

```python
df1
```

```python
# Number of stocks added each year.
yearly_counts = df1['Year'].value_counts().sort_index()
```

```python
yearly_counts
```

```python
#Maximum number of additions in year from 2020
yearly_counts[(yearly_counts).index > 2020].idxmax()
```

```python
# Number of companies with atleast 20 years
cutoff = pd.Timestamp.now() - pd.DateOffset(years=20)
stock_20_years = df1[df1['Date'] <= cutoff]
```
